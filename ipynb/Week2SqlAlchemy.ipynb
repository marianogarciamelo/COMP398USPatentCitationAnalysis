```python
inspector = inspect(engine)
existing_tables = inspector.get_table_names()

parquets_to_convert = list(Path('../parquets').glob('*.parquet'))

for parquet in parquets_to_convert:
    tableName = parquet.stem
    if tableName not in existing_tables:
        print(f"Attempting to read: {parquet}")
        df = pd.read_parquet(parquet)
        df = pd.read_parquet(parquet) ##need to read parquet in chunks
        print(f"converting {tableName} to SQL")
        df.to_sql(tableName, engine, if_exists='replace', index=False, method='multi', chunksize=1000)
        print(f"{tableName} converted to SQL")
    else:
        print(f"{tableName} already exists, skipping")
```
